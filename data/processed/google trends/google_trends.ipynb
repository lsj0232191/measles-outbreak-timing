```python
import pandas as pd                        
from pytrends.request import TrendReq
pytrend = TrendReq()
pytrend.build_payload(kw_list=['Measles'])
```

```python
related_topic = pytrend.related_queries()
related_topic.values()
```

```python
keywords = related_topic['Measles']['top']
keywords = keywords['query'].tolist()
for word in keywords:
    pytrend.build_payload(kw_list=[word])
    # all 50 states
    for state in ['US-AL', 'US-AK', 'US-AZ', 'US-AR', 'US-CA', 'US-CO', 'US-CT', 'US-DE', 'US-FL', 'US-GA', 'US-HI', 'US-ID', 'US-IL', 'US-IN', 'US-IA', 'US-KS', 'US-KY', 'US-LA', 'US-ME', 'US-MD', 'US-MA', 'US-MI', 'US-MN', 'US-MS', 'US-MO', 'US-MT', 'US-NE', 'US-NV', 'US-NH', 'US-NJ', 'US-NM', 'US-NY', 'US-NC', 'US-ND', 'US-OH', 'US-OK', 'US-OR', 'US-PA', 'US-RI', 'US-SC', 'US-SD', 'US-TN', 'US-TX', 'US-UT', 'US-VT', 'US-VA', 'US-WA', 'US-WV', 'US-WI', 'US-WY']:
        df_state = pytrend.get_historical_interest(word, year_start=2004, month_start=1, day_start=1, hour_start=0, year_end=2024, month_end=6, day_end=1, hour_end=0, cat=0, geo=state, gprop='', sleep=60)
        df_state.to_csv(f"../data/processed/google_trends/{word.replace(' ', '_')}_{state}.csv")
```
